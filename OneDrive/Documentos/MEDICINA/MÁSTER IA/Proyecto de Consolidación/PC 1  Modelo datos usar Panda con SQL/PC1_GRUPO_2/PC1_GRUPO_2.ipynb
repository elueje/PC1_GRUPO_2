```python
#Ahora vamos a confirmar la existencia de atípicos empleando el rango intercuartílico, recordemos que valor atípico leve es el que está fuera del intervalo media +-1.5std. 
#Posteriormente los eliminaremos, aunque no siempre es necesario eliminarlos porque en ocasiones nos pueden llegar a servir. 

import numpy as np

def remove_outliers_iqr(df):
    df_vino_completo_sin_atipicos = df.copy()  # Creamos una copia del dataframe original
    
    for col in df_vino_completo_sin_atipicos.select_dtypes(include=["float64", "int64"]).columns:  # Solo columnas numéricas
        Q1 = np.percentile(df_vino_completo_sin_atipicos[col], 25)  # Primer cuartil (25%)
        Q3 = np.percentile(df_vino_completo_sin_atipicos[col], 75)  # Tercer cuartil (75%)
        IQR = Q3 - Q1  # Rango intercuartil

        # Definir los límites inferior y superior
        lower_bound = Q1 - (1.5 * IQR)
        upper_bound = Q3 + (1.5 * IQR)

        # Filtrar datos eliminando los valores atípicos
        df_vino_completo_sin_atipicos = df_vino_completo_sin_atipicos[(df_vino_completo_sin_atipicos[col] >= lower_bound) & (df_vino_completo_sin_atipicos[col] <= upper_bound)]
    
    return df_vino_completo_sin_atipicos

# Aplicamos la función al DataFrame
df_vino_completo_sin_atipicos = remove_outliers_iqr(df_vino_completo)

print (df_vino_completo_sin_atipicos)
```

```python
# Ver cuántas filas eliminamos
print(f"Filas antes de eliminar atipicos: {df_vino_completo.shape[0]}")
print(f"Filas después de eliminar atipicos: {df_vino_completo_sin_atipicos.shape[0]}")
```

```python
#Vemos que hemos eliminado mucho valores atípicos , por lo tanto, vamos a retomar la función previa "df_vino_completo" y vamos a ver cuáles son los atípicos para intentar entender por qué se han generado losoutliers y así poder decidir la mejor manera de tratarlos. 
#Dado que han sido muchos, podemos sustituirlos por la mediana, aunque vamos a distorsionar la distribución. En caso de sustituirlos por valoers extremos, dados que son muchos los datos atípicos, podemos alterar la distribución. 

import pandas as pd
import numpy as np

# Seleccionamos únicamente las columnas numéricas
df_numerico = df_vino_completo.select_dtypes(include=['float64', 'int64'])

# Calcular IQR o rango intercuartílico
Q1 = df_numerico.quantile(0.25)
Q3 = df_numerico.quantile(0.75)
IQR = Q3 - Q1  

# Definir límites para detectar outliers
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Identificar valores atípicos
outliers = (df_numerico < limite_inferior) | (df_numerico > limite_superior)

# Mostrar las filas que contienen valores atípicos y qué valores son
df_atipicos = df_numerico[outliers]

# Imprimir los valores atípicos
print("Valores atípicos detectados:")
print(df_atipicos.dropna(how='all'))  # Eliminamos filas donde no hay outliers

```

```python
#Tras analizar los datos atípicos, son muchos datos que no se encuentran disponibles de manera aleatoria, por lo que considero que lo más indicados es suprimir los datos atípicos. 
#Aclarar que en el DF previo, los valores que aparecen como NaN no son datos atípicos, sino que se encuentran dentro de los límites que hemos considerado como nrormales, es decir, dentro del IQR definido. 

import numpy as np

def remove_outliers_iqr(df):
    df_vino_completo_sin_atipicos = df.copy()  # Creamos una copia del dataframe original
    
    for col in df_vino_completo_sin_atipicos.select_dtypes(include=["float64", "int64"]).columns:  # Solo columnas numéricas
        Q1 = np.percentile(df_vino_completo_sin_atipicos[col], 25)  # Primer cuartil (25%)
        Q3 = np.percentile(df_vino_completo_sin_atipicos[col], 75)  # Tercer cuartil (75%)
        IQR = Q3 - Q1  # Rango intercuartil

        # Definir los límites inferior y superior
        lower_bound = Q1 - (1.5 * IQR)
        upper_bound = Q3 + (1.5 * IQR)

        # Filtrar datos eliminando los valores atípicos
        df_vino_completo_sin_atipicos = df_vino_completo_sin_atipicos[(df_vino_completo_sin_atipicos[col] >= lower_bound) & (df_vino_completo_sin_atipicos[col] <= upper_bound)]
    
    return df_vino_completo_sin_atipicos

# Aplicamos la función al DataFrame
df_vino_completo_sin_atipicos = remove_outliers_iqr(df_vino_completo)

print (df_vino_completo_sin_atipicos)
```

```python
# Ejercicio 4. Almacenar los datos limpios en SQLite.
import sqlite3

# Crear la conexión a SQLite
conexion = sqlite3.connect("vino_limpio.db")  # Base de datos SQLite
cursor = conexion.cursor()

# Crear la tabla (si no existe)
cursor.execute('''
    CREATE TABLE IF NOT EXISTS vinos (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        fixed_acidity REAL,
        volatile_acidity REAL,
        citric_acid REAL,
        residual_sugar REAL,
        chlorides REAL,
        free_sulfur_dioxide REAL,
        total_sulfur_dioxide REAL,
        density REAL,
        pH REAL,
        sulphates REAL,
        alcohol REAL,
        quality INTEGER,
        tipo_de_vino TEXT
    )
''')

conexion.commit() # Guardamos los cambios

df_vino_completo_sin_atipicos.to_sql("vinos", conexion, if_exists="replace", index=False) # Insertar los datos en la tabla

# Verificar que se han insertado correctamente
cursor.execute("SELECT COUNT(*) FROM vinos")
num_registros = cursor.fetchone()[0]
print(f"Se han insertado {num_registros} registros en la base de datos.")

# Cerrar conexión
conexion.close()
```
