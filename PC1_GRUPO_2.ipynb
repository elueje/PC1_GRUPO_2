```python
# Usa Pandas para cargar ambos datasets en dataframes en memoria.

import pandas as pd

# Me descargo y guardo los datasets en una carpeta que llamo data dentro de PC1_GRUPO_2 "PC1_GRUPO_2/data/"
df_red = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv", sep=";")
df_white = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv", sep=";")

print("Archivos descargados y guardados correctamente en PC1_proyecto/data/")
```

```python
# Añade una columna adicional que indique el tipo de vino (`red` o `white`). 

df_red["tipo_de_vino"] = "tinto"

df_red.head()
```

```python
df_white["tipo_de_vino"] = "blanco"

df_white.head()
```

```python
# Combina los datos en un único dataframe con la nueva columna.

df_vino_tinto_y_blanco = pd.concat([df_red, df_white])

df_vino_tinto_y_blanco.head
```
